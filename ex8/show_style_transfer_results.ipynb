```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt


STYLE_DIR = Path("./style_img")
OUTPUT_DIR = Path("./output_img")
CONTENT_DIR = Path("./cat.png")


items = [
    {
        "style_file": "source_code_style_1.jpg",
        "output_file": "result_1.png",
        "model_text": "Source code model",
    },
    {
        "style_file": "source_code_style_2.jpg",
        "output_file": "result_2.png",
        "model_text": "Source code model",
    },
    {
        "style_file": "source_code_style_3.jpg",
        "output_file": "result_3.png",
        "model_text": "Source code model",
    },
    {
        "style_file": "our_style_img.png",
        "output_file": "result_our.png",
        "model_text": "Our trained model",
    },
]


def load_image(path):
    return Image.open(path).convert("RGB")


def show_cell(ax, image_path=None, title="", title_size=18):
    if image_path is not None:
        ax.imshow(load_image(image_path))

    ax.set_title(title, fontsize=title_size, fontweight="bold", pad=14, color="#FF5733")
    ax.set_xticks([])
    ax.set_yticks([])

    for spine in ax.spines.values():
        spine.set_visible(False)


fig, axes = plt.subplots(
    2,
    len(items) + 1,
    figsize=(28, 11),
    gridspec_kw={
        "wspace": 0.0,
        "hspace": 0.18,
    },
)

# 左上は空欄
show_cell(
    axes[0, 0],
    title="",
)

# 2行目・左端に Original Image
show_cell(
    axes[1, 0],
    CONTENT_DIR,
    "Original Image",
    title_size=20,
)

# 1行目: Style images
# 2行目: Stylized results
for i, item in enumerate(items, start=1):
    show_cell(
        axes[0, i],
        STYLE_DIR / item["style_file"],
        f"Style {i}\n{item['model_text']}",
        title_size=16,
    )

    show_cell(
        axes[1, i],
        OUTPUT_DIR / item["output_file"],
        f"Stylized Result {i}",
        title_size=18,
    )
plt.show()
```
